## 3. Conceptual Foundations and Formal Semantics

### 1. Python names reference objects

`x = something` binds a name to an object; it does not establish a C-style typed value slot. Python lists are resizable containers of references, each of which may point to objects with independent type and allocation histories. Under common CPython implementations, this creates indirection and object-management overhead, but exact byte overhead is platform and implementation dependent.

### 2. The NumPy array object

A normal numeric `ndarray` stores elements according to one dtype descriptor within a data buffer. Numeric array elements need not be individually allocated Python integers or floats. Python-level indexing that extracts a scalar can create a NumPy scalar object, but that does not imply the underlying array consisted of individually allocated scalar objects.

### 3. Identity, aliasing, and mutation

`a is b` asks whether names reference the same Python object. `np.shares_memory(a, b)` asks whether the arrays' memory regions overlap, which can hold for distinct array objects. Value equality (`np.array_equal`) is a third independent property. An object may be different yet alias the same allocation, or store equal values in independent allocations.

### 4. Object dtype exception

`dtype=object` keeps a homogeneous descriptor at the array level, but its elements contain Python object references. Such arrays do not exhibit the same numerical-kernel and memory characteristics as packed numeric arrays. They may require the GIL for object operations and inherit ordinary Python semantics.

### 5. Behavioral contracts

Lists and arrays differ in `+`, `*`, equality, multidimensional indexing, and elementwise comparisons. Code should never rely on superficial similarity in display syntax. A robust API explicitly specifies whether it consumes Python sequences, NumPy arrays, or arbitrary ArrayLike values.

## 5. Executable Investigations

Each investigation is independent in purpose; execute in sequence so shared imports are available.

### 5.1 — Objects and numeric storage

**Research question.** What does this experiment establish, and which conclusion would be unjustified if we only inspected its printed values?

```python
import numpy as np, sys
values = list(range(1000))
a = np.asarray(values, dtype=np.int64)
print('list container bytes:', sys.getsizeof(values))
print('array element bytes:', a.nbytes, 'array object bytes:', sys.getsizeof(a))
print('Note: list container size excludes separate Python int objects.')
```

### 5.2 — Identity, storage overlap, equality

**Research question.** What does this experiment establish, and which conclusion would be unjustified if we only inspected its printed values?

```python
a = np.arange(6)
b = a[1:5]
c = b.copy()
assert a is not b
assert np.shares_memory(a, b)
assert not np.shares_memory(a, c)
assert np.array_equal(b, c)
b[0] = -10
assert a[1] == -10 and c[0] == 1
```

### 5.3 — Object dtype is not packed numeric representation

**Research question.** What does this experiment establish, and which conclusion would be unjustified if we only inspected its printed values?

```python
packed = np.array([1, 2, 3], dtype=np.int64)
objects = np.array([1, 2, 3], dtype=object)
print('packed itemsize:', packed.itemsize, 'object reference slot:', objects.itemsize)
assert objects.dtype == object
assert packed.dtype != object
```

## 6. Edge Cases, Failure Modes, and Misconceptions

- **Logical validity is not physical contiguity.** Shape alone cannot determine whether an operation needs a copy.
- **Value equality is not identity.** Two arrays may compare equal while having independent storage, or share storage while being distinct Python objects.
- **A vectorized expression does not certify correctness.** Broadcasting and axis conventions can silently encode unintended mathematics.
- **A dtype is not a real-number model.** Overflow, finite precision, and representation-dependent transformations require explicit attention.
- **A benchmark does not imply a theorem.** Hardware, build configuration, allocation strategy, and input scale can affect conclusions.
- **Array semantics do not encode domain assumptions.** For finance or statistics, look-ahead restrictions, missing-data rules, weights, and alignment remain external responsibilities.

When evaluating an unexpected result, first inspect `shape`, `dtype`, `strides`, and storage overlap, then isolate the mathematical operation on a hand-checkable example.

## 7. Performance and Complexity Considerations

The performance question should be asked in three parts: **work complexity**, **data movement**, and **peak live memory**. A contiguous numeric array can support efficient native processing, but a particular operation may be limited by memory bandwidth, temporary allocation, scalar dispatch, or native-library parallelism. Use measurements only after verifying correctness, and never assume that a microbenchmark generalizes to different dtypes or problem sizes.

A practical benchmark should (i) isolate the operation, (ii) distinguish input setup from computation, (iii) warm up the relevant execution path, (iv) collect repeated measurements, and (v) record platform and NumPy provenance.

## 8. Analytical Interpretation

A useful separation is: (1) *name identity*, (2) *container identity*, (3) *storage identity*, and (4) *value equality*. Confusing them causes hidden mutations in model pipelines. Storage efficiency arises from specialized representation, not from an assertion that “NumPy objects have no overhead.”
