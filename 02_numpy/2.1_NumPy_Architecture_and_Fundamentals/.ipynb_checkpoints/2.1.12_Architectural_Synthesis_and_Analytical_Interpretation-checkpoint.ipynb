## 3. Conceptual Foundations and Formal Semantics

### 1. A unified abstraction

A NumPy computation can be decomposed into a **logical specification** (shape and mathematical transformation), a **representation contract** (dtype, strides, ownership), a **dispatch strategy** (ufunc, library call, subclass override), and a **validation layer** (shape, numerical tolerances, side effects, and application assumptions). Each layer can fail independently.

### 2. Case study: multi-asset scenario returns

Let `R` have shape `(S,T,N)` for scenario, time, asset. Let `w` have shape `(N,)` and represent constant asset weights. Then the portfolio return tensor is `p_{s,t} = Σ_j R_{s,t,j} w_j`. NumPy can compute it with `R @ w`. This equation does not imply that compounded wealth is additive; for compounded returns, compute a product of `(1+p)` along the time axis under the chosen rebalancing assumption.

### 3. Storage and execution review

The input may be a view, the weight vector may be broadcast, and `R @ w` will typically produce a new array. A model with millions of scenarios may be limited by reads from memory, not multiplication count. The application should record dtype policy, shape conventions, NaN policy, and benchmark characteristics.

### 4. Verification hierarchy

Begin with a tiny hand-checkable example. Assert shapes and elementwise identities. Verify absence of unintended mutation. Compare alternative formulations under tolerances. Test degenerate shapes, invalid dtypes, and non-finite data. Only then benchmark scaling on realistic workloads.

### 5. Architectural decision record

For every major operation, record: intended mathematical contract, array shape and dtype, mutation policy, expected output shape, numerical tolerance, complexity and memory budget, and environmental reproducibility requirements. This lightweight discipline prevents many subtle scientific errors.

## 4. Architectural Model and Invariants

A useful four-layer checklist for this topic:

| Layer | Question to ask | Representative invariant |
|---|---|---|
| Logical | What operation is mathematically intended? | Axes and result shape correspond to the model |
| Representation | What dtype and storage model represent the inputs? | No unsupported casting or unexpected aliasing |
| Execution | Which category of operation processes the arrays? | Results are stable under a reference implementation |
| Validation | What independent evidence establishes correctness? | Assertions, boundary tests, and numerical tolerances |

**Important:** Shape agreement and code execution are not sufficient evidence of mathematical validity. Conversely, a mathematically valid expression can be operationally impractical or numerically unstable.

## 5. Executable Investigations

Each investigation is independent in purpose; execute in sequence so shared imports are available.

### 5.1 — Scenario/time/asset portfolio model

**Research question.** What does this experiment establish, and which conclusion would be unjustified if we only inspected its printed values?

```python
import numpy as np
R = np.array([[[0.02,-0.01],[0.01,0.02]], [[-0.01,0.03],[0.04,-0.02]]]) # (S,T,N)
w = np.array([0.6,0.4])
p = R @ w
assert p.shape == (2,2)
np.testing.assert_allclose(p[0], [0.008,0.014])
wealth = np.prod(1+p, axis=1)
print('scenario terminal wealth:', wealth)
```

### 5.2 — Independent cross-check and purity assertion

**Research question.** What does this experiment establish, and which conclusion would be unjustified if we only inspected its printed values?

```python
R_original = R.copy()
reference = np.einsum('stn,n->st', R, w)
np.testing.assert_allclose(p, reference)
np.testing.assert_array_equal(R, R_original)
assert not np.shares_memory(p, R)
```

### 5.3 — Boundary and failure contracts

**Research question.** What does this experiment establish, and which conclusion would be unjustified if we only inspected its printed values?

```python
def portfolio_returns(returns, weights):
    r = np.asarray(returns)
    w = np.asarray(weights)
    if r.ndim != 3 or w.ndim != 1 or r.shape[-1] != w.shape[0]:
        raise ValueError('Expected returns (scenario,time,asset) and weights (asset,)')
    if not (np.issubdtype(r.dtype, np.number) and np.issubdtype(w.dtype, np.number)):
        raise TypeError('Numeric dtypes required')
    if not (np.all(np.isfinite(r)) and np.all(np.isfinite(w))):
        raise ValueError('All inputs must be finite')
    return r @ w
np.testing.assert_allclose(portfolio_returns(R,w),p)
try:
    portfolio_returns(R, np.ones(3))
except ValueError:
    print('shape contract correctly rejected invalid weights')
else:
    raise AssertionError('expected failure')
```

## 6. Edge Cases, Failure Modes, and Misconceptions

- **Logical validity is not physical contiguity.** Shape alone cannot determine whether an operation needs a copy.
- **Value equality is not identity.** Two arrays may compare equal while having independent storage, or share storage while being distinct Python objects.
- **A vectorized expression does not certify correctness.** Broadcasting and axis conventions can silently encode unintended mathematics.
- **A dtype is not a real-number model.** Overflow, finite precision, and representation-dependent transformations require explicit attention.
- **A benchmark does not imply a theorem.** Hardware, build configuration, allocation strategy, and input scale can affect conclusions.
- **Array semantics do not encode domain assumptions.** For finance or statistics, look-ahead restrictions, missing-data rules, weights, and alignment remain external responsibilities.

When evaluating an unexpected result, first inspect `shape`, `dtype`, `strides`, and storage overlap, then isolate the mathematical operation on a hand-checkable example.

## 7. Performance and Complexity Considerations

The performance question should be asked in three parts: **work complexity**, **data movement**, and **peak live memory**. A contiguous numeric array can support efficient native processing, but a particular operation may be limited by memory bandwidth, temporary allocation, scalar dispatch, or native-library parallelism. Use measurements only after verifying correctness, and never assume that a microbenchmark generalizes to different dtypes or problem sizes.

A practical benchmark should (i) isolate the operation, (ii) distinguish input setup from computation, (iii) warm up the relevant execution path, (iv) collect repeated measurements, and (v) record platform and NumPy provenance.

## 8. Analytical Interpretation

NumPy competence is not memorizing functions; it is maintaining multiple simultaneous invariants. A *correct-looking* array does not prove correct axis meaning, no aliasing, numerical stability, or economic validity. The expert workflow explicitly verifies each layer and documents why the chosen numerical abstraction fits the scientific question.
