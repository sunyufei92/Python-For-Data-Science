## 5. Difficulty calibration

**Level 1 — Foundation.** Correct knowledge of interfaces and straightforward operations; elementary complexity.

**Level 2 — Intermediate.** Combines several operations; explains common failure modes, shapes and edge cases.

**Level 3 — Advanced.** Requires algorithmic design, rigorous complexity, testing, and meaningful implementation tradeoffs.

**Level 4 — Expert.** Requires explicit memory reasoning, numerical or statistical rigor, scalable alternatives, and nuanced failure analysis.

**Level 5 — Research / production challenge.** Solves problems under real constraints: streaming data, limited memory, leakage prevention, distributed computation, stability, and monitoring.

**Calibration rules:**
- Difficulty is judged by the *reasoning required*, not by the length of the solution or obscurity of syntax.
- A question can be easy for one role and difficult for another; specify the target role where material.
- Do **not** equate company brand with difficulty.
- Tag `time_limit_min` only when a source actually reports one, or label it as a proposed practice limit.

## 9. Sample entry (illustrative — **not** a reported company question)

**ID:** `NUMPY-0001` · **Difficulty:** Level 4 (Expert) · **Type:** `code_prediction`, `complexity`, `numerical_reasoning`

**Interview prompt:**

Given arrays `X ∈ ℝ^(n×d)` and `Y ∈ ℝ^(m×d)`, analyze:

```python
squared = ((X[:, None, :] - Y[None, :, :]) ** 2).sum(axis=-1)
```

1. Determine the shapes of the broadcast operands, intermediate result, and output.
2. Derive computational complexity and **peak temporary memory** under standard eager NumPy semantics.
3. Explain why NumPy broadcasting avoids explicitly replicating the inputs yet **does not** prevent allocation of the arithmetic result.
4. Propose a blocked alternative under a strict memory limit.
5. Discuss why an identity based on `||x||² + ||y||² − 2x·y` may experience catastrophic cancellation for nearly equal vectors.

The next code cell performs a small **shape sanity check**, not a full reference solution.

````python
import numpy as np

rng = np.random.default_rng(2026)
X = rng.normal(size=(4, 3))
Y = rng.normal(size=(5, 3))

a = X[:, None, :]
b = Y[None, :, :]
diff = a - b
squared = np.sum(diff ** 2, axis=-1)

assert a.shape == (4, 1, 3)
assert b.shape == (1, 5, 3)
assert diff.shape == (4, 5, 3)
assert squared.shape == (4, 5)
assert np.all(squared >= 0)

print('X expanded:', a.shape)
print('Y expanded:', b.shape)
print('Broadcast result:', diff.shape)
print('Pairwise squared distances:', squared.shape)
````

**Analytical interpretation.** For `float64`, `diff` contains `n × m × d` elements and therefore uses approximately `8nmd` bytes of *element storage*. The squaring and reduction steps may allocate additional temporary buffers (and the output is `8nm` bytes); exact **peak** resident memory is implementation- and expression-dependent. A `n=100_000, m=50_000, d=128` pairwise tensor alone would need approximately **5.12 TB** of element storage (decimal TB). Therefore, for large inputs, reformulate or block the computation; a full dense `n × m` float64 result alone needs **40 GB**.

A robust expert answer distinguishes **input storage**, **broadcast views**, **temporary allocations**, **output storage**, and **floating-point error** rather than saying “broadcasting uses no memory.”

## 11. Practice session tracker (runnable)

This lightweight example demonstrates a **local, in-memory** registry for practice attempts. It does not claim to save progress between notebook sessions; use a dedicated progress-tracker notebook or export records explicitly if persistence is needed.

````python
from dataclasses import dataclass
from typing import Literal

Status = Literal["unsolved", "partial", "solved"]

@dataclass(frozen=True)
class InterviewAttempt:
    question_id: str
    domain: str
    level: int                 # 1 (Foundation) to 5 (Research/Production)
    status: Status
    minutes_spent: float
    correctness_score: float  # 0 to 100

    def __post_init__(self):
        if not self.question_id.strip() or not self.domain.strip():
            raise ValueError("question_id and domain must be nonempty")
        if not 1 <= self.level <= 5:
            raise ValueError("level must be between 1 and 5")
        if self.minutes_spent < 0:
            raise ValueError("minutes_spent cannot be negative")
        if not 0 <= self.correctness_score <= 100:
            raise ValueError("correctness_score must be in [0, 100]")

attempts = [
    InterviewAttempt("NUMPY-0001", "NumPy", 4, "partial", 32, 70),
    InterviewAttempt("ALGO-0001", "Algorithms", 3, "solved", 21, 95),
    InterviewAttempt("STAT-0001", "Statistics", 4, "unsolved", 35, 30),
]

for attempt in sorted(attempts, key=lambda a: (a.correctness_score, -a.level)):
    print(f"{attempt.question_id:<12} {attempt.status:<9} "
          f"score={attempt.correctness_score:>5.1f} "
          f"time={attempt.minutes_spent:.0f}m")
````
