## Overview

This notebook is a structured collection of **Pandas technical interview questions** focused on solving realistic data analysis and data manipulation problems under interview conditions. It emphasizes accurate reasoning, executable Python solutions, interpretation of results, and the ability to explain performance and correctness trade-offs.

The question bank is organized into **12 interview-oriented knowledge areas**. These areas serve as a classification system for company-reported questions and original advanced practice problems, rather than as chapters of a general Pandas tutorial.

Where credible public records exist, the notebook will include questions reported by candidates interviewing at companies such as Google, Meta, Amazon, Microsoft, Netflix, Bloomberg, and financial institutions. Company attribution is recorded only when supported by a traceable source. An adapted exercise must never be presented as an exact historical interview question.

**Guiding principle:** One question may assess multiple areas of Pandas. Each question has one *primary category* and may have several *secondary topic tags*, allowing questions to be organized without unnecessary duplication.

## 1. Learning Objectives

After working through this notebook, a candidate should be able to:

1. Explain the behavior of Pandas `Series`, `DataFrame`, indexes, and data types in interview-level detail.
2. Select, filter, align, and transform structured data without introducing subtle correctness errors.
3. Construct aggregation, merging, reshaping, ranking, and window-computation pipelines.
4. Handle missing values, duplicates, outliers, and invalid records using defensible data-quality rules.
5. Solve time-series questions while respecting event ordering, timestamp semantics, and leakage constraints.
6. Compare alternative implementations in terms of readability, computational complexity, memory usage, and scalability.
7. Translate loosely specified analytical tasks into explicit input/output requirements, edge cases, and testable solutions.
8. Communicate solutions, technical trade-offs, and analytical implications clearly in an interview.

The objective is **not memorization of Pandas syntax**. It is the ability to solve unfamiliar, multi-step data problems correctly and efficiently.

### Classification Rules

- Assign **one primary category** representing the dominant competency tested.
- Add **secondary tags** for other relevant concepts; do not duplicate the full question in several sections.
- Reserve **Category 12** for multi-stage analytical problems in which the main challenge is combining several techniques, rather than demonstrating a single API.
- Preserve overlaps intentionally: for example, `MultiIndex` may appear in Categories 02 and 08, and vectorization in Categories 03 and 11. Classify by the *reasoning objective*, not merely by which function appears in the code.
- The table establishes **coverage areas**, not a quota or claimed empirical distribution of questions across companies.

## 3. Source and Authenticity Standards

This notebook distinguishes carefully between **what a candidate or a company actually reported** and **what has been reconstructed for practice**.

| Source status | Meaning | Attribution rule |
|---|---|---|
| **Reported Interview Question** | A question or task described in a traceable public interview report | Record company, role, report date or interview year when known, and source URL; make clear when the wording is paraphrased |
| **Adapted Interview Question** | A modified, expanded, or executable version derived from a reported interview topic | Cite the underlying report and explicitly identify any invented dataset, constraints, or follow-up tasks |
| **Original Practice Question** | A newly created exercise designed to assess interview-relevant skills | Do **not** attribute to a specific company without a supporting source |

**Evidence policy**

1. Do not label a question as a historical company question based solely on its style, difficulty, or similarity to another exercise.
2. Treat anonymous candidate accounts as **self-reported**, not as officially confirmed company exam records.
3. Do not infer an interview year from a webpage's update date; distinguish the **interview date** from the **publication date** whenever possible.
4. If an interview report states only a broad topic (e.g., *grouping and merging*), do not invent an exact original prompt. Label a newly specified exercise **Adapted**.
5. Record uncertain or unavailable details as **Not reported** rather than guessing.
6. Prefer direct reports and archived source URLs over unattributed question compilations. Preserve links for later verification.

These rules support a question bank that remains auditable as new interview reports are added.

Difficulty is a **notebook-specific assessment**. It is not an official grading system used by the named companies.

A short question can still be difficult when it demands precise reasoning about index alignment, missing values, copies versus views, or temporal leakage.

## 5. Standard Format for Each Interview Question

Questions will be added **one at a time**. The following structure should be used consistently, with sections omitted only when genuinely inapplicable.

### Question `PD-0001` — *Question Title*

**Metadata**

| Field | Value |
|---|---|
| Question ID | `PD-0001` |
| Primary category | `04 — GroupBy and Aggregation` |
| Secondary tags | e.g., `merge`, `filter`, `sort_values` |
| Difficulty | Level 1–5 |
| Company / position | Named only if supported |
| Interview year | Year, or `Not reported` |
| Source status | Reported / Adapted / Original Practice |
| Source URL | Direct link, or `Not applicable` |

**1. Interview Question** — The question as reported, paraphrased, or authored, with its status clearly stated.

**2. Problem Requirements** — Inputs, expected output, constraints, assumptions, and important edge cases.

**3. Conceptual Analysis** — Required Pandas semantics and a sound approach to the problem.

**4. Python Implementation** — Executable code in separate **Code cells**, using self-contained data and explicit variable names.

**5. Expected Results and Validation** — Reference output, assertions, and edge-case checks.

**6. Complexity and Performance Analysis** — Time complexity, memory behavior, and relevant Pandas execution considerations. Avoid asserting precise complexity for library internals without qualification.

**7. Alternative Approaches** — Compare clarity, performance, numerical behavior, or robustness where relevant.

**8. Common Pitfalls** — Explain errors such as unintended index alignment, ambiguous merges, `NaN` propagation, and accidental copies.

**9. Interview Follow-ups** — Stronger constraints, scale-up questions, or changes to the original problem.

**10. Analytical Interpretation** — Explain what the result means for the underlying dataset or business/statistical question; acknowledge limitations.

> **Cell convention:** Use Markdown cells for the problem and reasoning; use Code cells for imports, reproducible test data, solutions, and checks. Avoid placing large runnable solutions inside Markdown code fences.

## 6. Solution Quality and Evaluation Criteria

Every completed answer should be evaluated along these dimensions:

| Dimension | Questions to ask |
|---|---|
| **Correctness** | Is the output right for standard, empty, duplicated, and missing-data cases? |
| **Semantic precision** | Are index alignment, types, null handling, grouping keys, and ordering explicit? |
| **Computational efficiency** | Are there avoidable Python-level loops, repeated merges, unnecessary copies, or large intermediate objects? |
| **Robustness** | Are assumptions validated and failure modes identified? |
| **Communication** | Can the candidate explain the approach and trade-offs without depending on trial and error? |
| **Analytical validity** | Does the transformation answer the intended question without leakage or incorrect denominators? |
| **Reproducibility** | Can the question be run independently and checked with assertions? |

**Note:** A solution is not automatically superior just because it uses fewer lines or is fully vectorized. Clarity, correctness, resource use, and the data's size and structure all matter.

## 7. Notebook Conventions and Workflow

- **Language:** Interview questions, explanations, and technical discussion are written in English.
- **Question order:** Add questions sequentially using stable IDs (`PD-0001`, `PD-0002`, …). Their topic category may differ from the numerical order.
- **Source annotations:** Retain the exact URL and explicitly mark adapted elements.
- **Dependencies:** Prefer `pandas` and `numpy` unless a question requires another library. Record assumptions if a result depends on a particular version.
- **Reproducibility:** Use compact synthetic datasets; fix random seeds where randomness matters.
- **Validation:** Include expected results and checks, rather than relying only on visually inspecting a printed table.
- **Independence:** Whenever practical, each question should run without relying on hidden state left by previous questions.
- **Incremental maintenance:** Add one complete question and its solution before proceeding to the next.

### Coverage Tracking

The 12 categories are **tags and organizational anchors**, not a mandatory writing order. A real company question may be recorded as soon as its source and requirements are sufficiently clear, then assigned to the appropriate category.
