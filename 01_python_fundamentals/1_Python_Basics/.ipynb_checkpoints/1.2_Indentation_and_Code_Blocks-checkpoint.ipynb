# 1.2 Indentation and Code Blocks

## Overview

Indentation is a fundamental component of Python's syntax. Unlike programming languages that use braces to delimit code blocks, Python uses whitespace to determine the structure and execution flow of a program.

Understanding indentation is essential not only for writing syntactically correct code but also for implementing complex algorithms, nested decision structures, exception handling, and financial computation pipelines.

Incorrect indentation can produce syntax errors or, more dangerously, logically incorrect results without raising an exception.

### Learning Objectives

By the end of this section, you will be able to:

1. Understand how Python defines and executes code blocks.
2. Implement complex nested control structures.
3. Distinguish block indentation from expression continuation.
4. Understand the relationship between indentation and variable scope.
5. Identify syntax errors and logical errors caused by incorrect indentation.
6. Design structured, readable, and maintainable analytical programs.
7. Apply advanced control-flow structures to quantitative portfolio risk analysis.


## 1.2.1 Python Code Blocks and Suite Structure

### Definition

A code block, also called a *suite* in Python's grammar, is a group of statements associated with a compound statement.

Compound statements include:

- Conditional statements: `if`, `elif`, `else`
- Iteration statements: `for`, `while`
- Function definitions: `def`
- Class definitions: `class`
- Exception handling: `try`, `except`, `finally`
- Context managers: `with`

### Fundamental Rules

1. A colon (`:`) introduces a suite in a compound statement.
2. Indentation determines which statements belong to the same suite.
3. Four spaces per indentation level is the recommended convention.
4. Statements at the same indentation level belong to the same enclosing block.
5. Dedentation indicates the end of an indented block.
6. Inconsistent indentation can raise an `IndentationError` or `TabError`.

### Logical Significance

Indentation is part of Python's program semantics.

Changing the indentation of a statement can change when it executes, even when the program remains syntactically valid.


```python
# Portfolio performance monitoring

asset_returns = {
    "AAPL": -0.12,
    "MSFT": 0.04,
    "JPM": -0.07,
    "NVDA": -0.19,
    "TLT": 0.03
}

warning_threshold = -0.05
critical_threshold = -0.15

risk_alerts = []

for ticker, asset_return in asset_returns.items():

    if asset_return <= critical_threshold:
        severity = "CRITICAL"

    elif asset_return <= warning_threshold:
        severity = "WARNING"

    else:
        severity = "NORMAL"

    if severity != "NORMAL":
        risk_alerts.append({
            "ticker": ticker,
            "return": asset_return,
            "severity": severity
        })

# Outside the loop: executed once
print("PORTFOLIO RISK MONITORING")
print("-" * 45)

for alert in risk_alerts:
    print(
        f"{alert['ticker']:<6}"
        f"{alert['return']:>9.2%}  "
        f"{alert['severity']}"
    )

print(f"\nTotal Alerts: {len(risk_alerts)}")
```

### Execution Flow Analysis

The program contains multiple levels of indentation:

- **Level 0:** Global configuration and report generation.
- **Level 1:** Statements executed for each asset.
- **Level 2:** Statements executed within conditional branches and nested structures.

The `for` loop evaluates each asset independently.

The conditional branches determine the severity classification, while the subsequent conditional statement decides whether an alert should be recorded.

The final report is generated after the loop terminates.

**Important:** If the report-generation code were incorrectly indented inside the main loop, it would execute repeatedly, potentially producing incomplete intermediate reports.

This illustrates why indentation affects program behavior rather than merely visual formatting.


## 1.2.2 Nested Code Blocks and Complex Control Flow

### Nested Structures

Nested code blocks occur when a compound statement is placed inside another compound statement.

They are frequently used in:

- Hierarchical decision-making
- Multi-asset portfolio analysis
- Financial risk monitoring
- Data validation pipelines
- Iterative optimization algorithms
- Scenario-based simulations

### Execution Principles

1. Outer conditions determine whether inner blocks are reached.
2. Inner conditions are evaluated only when their enclosing blocks execute.
3. Each additional level of nesting increases structural complexity.
4. Excessive nesting can reduce readability and increase the likelihood of logical errors.

### Best Practices

- Keep nesting depth manageable.
- Use meaningful intermediate variables.
- Prefer early exits when appropriate.
- Extract complex logic into functions when it becomes difficult to follow.
- Separate validation, computation, and reporting logic whenever possible.


```python
# Hierarchical risk assessment

portfolio = {
    "AAPL": {
        "sector": "Technology",
        "weight": 0.28,
        "return": -0.12
    },
    "MSFT": {
        "sector": "Technology",
        "weight": 0.24,
        "return": 0.05
    },
    "JPM": {
        "sector": "Financials",
        "weight": 0.18,
        "return": -0.08
    },
    "XOM": {
        "sector": "Energy",
        "weight": 0.15,
        "return": 0.09
    },
    "TLT": {
        "sector": "Bonds",
        "weight": 0.15,
        "return": -0.02
    }
}

MAX_POSITION_WEIGHT = 0.25
LOSS_THRESHOLD = -0.10

risk_assessments = {}

for ticker, asset in portfolio.items():

    weight = asset["weight"]
    asset_return = asset["return"]

    if weight > MAX_POSITION_WEIGHT:

        if asset_return < LOSS_THRESHOLD:
            risk_status = "CRITICAL"

        elif asset_return < 0:
            risk_status = "HIGH"

        else:
            risk_status = "CONCENTRATED"

    else:

        if asset_return < LOSS_THRESHOLD:
            risk_status = "LOSS ALERT"

        elif asset_return < 0:
            risk_status = "MONITOR"

        else:
            risk_status = "NORMAL"

    risk_assessments[ticker] = risk_status

for ticker, status in risk_assessments.items():
    print(f"{ticker:<6} | {status}")
```

## 1.2.3 Indentation vs. Line Continuation

Python distinguishes between indentation that defines code blocks and indentation used for formatting expressions across multiple lines.

### Block Indentation

Block indentation is syntactically significant. It determines which statements belong to a compound statement.

### Continuation Indentation

Continuation indentation improves the readability of expressions that span multiple physical lines.

When an expression is enclosed in parentheses `()`, brackets `[]`, or braces `{}`, Python allows implicit line continuation.

Continuation indentation does not create an additional code block.

### Recommended Practices

- Use four spaces for block indentation.
- Use parentheses for long expressions.
- Align continuation lines consistently.
- Avoid excessive nesting in complex expressions.
- Prefer named intermediate calculations when expressions become difficult to interpret.


```python
# Portfolio configuration

equity_value = 85000.00
bond_value = 35000.00
cash_value = 15000.00

equity_shock = -0.18
bond_shock = -0.07

# Multi-line calculation using implicit continuation

portfolio_value = (
    equity_value
    + bond_value
    + cash_value
)

stressed_portfolio_value = (
    equity_value * (1 + equity_shock)
    + bond_value * (1 + bond_shock)
    + cash_value
)

stress_return = (
    stressed_portfolio_value / portfolio_value
    - 1
)

# Multi-line conditional expression

if (
    stress_return < -0.10
    and equity_value / portfolio_value > 0.50
):
    risk_classification = "HIGH RISK"

else:
    risk_classification = "ACCEPTABLE"

print(f"Portfolio Value: ${portfolio_value:,.2f}")
print(f"Stressed Value: ${stressed_portfolio_value:,.2f}")
print(f"Stress Return: {stress_return:.2%}")
print(f"Risk Classification: {risk_classification}")
```

## 1.2.4 Indentation and Variable Scope

### Block Structure vs. Variable Scope

Although indentation defines code blocks, not every indented block creates a new variable scope.

In Python, scope is primarily determined by functions, modules, classes, and enclosing function contexts.

Unlike some other programming languages, ordinary `if`, `for`, and `while` blocks do not introduce separate local scopes.

### Important Implications

- Variables assigned inside conditional blocks may remain accessible afterward.
- Variables assigned inside loops can remain accessible after loop termination.
- Variables defined inside functions are generally local to those functions.
- A variable assigned only in a conditional branch may remain undefined if that branch never executes.

### Common Risks

1. Uninitialized variables.
2. Unexpected reuse of variables.
3. Variable shadowing.
4. Incorrect assumptions about block-level scope.
5. Accidental dependencies on loop variables.

Reliable Python programs should initialize variables explicitly and avoid depending on uncertain execution paths.


```python
# Variables assigned inside conditional blocks

portfolio_return = -0.08

if portfolio_return < 0:
    market_condition = "Negative"
else:
    market_condition = "Positive"

# Accessible because if/else does not create a new scope
print(market_condition)


# Variables assigned inside loops

positions = {
    "AAPL": 100,
    "MSFT": 50,
    "JPM": 75
}

total_shares = 0

for ticker, shares in positions.items():
    total_shares += shares

print(f"Total Shares: {total_shares}")


# Function scope behaves differently

def calculate_exposure(market_value, portfolio_value):

    exposure = market_value / portfolio_value

    return exposure


result = calculate_exposure(25000, 100000)

print(f"Exposure: {result:.2%}")

# 'exposure' is local to calculate_exposure().
# Accessing it directly here would raise NameError.
```

## 1.2.5 Common Indentation Errors and Logical Pitfalls

### 1. IndentationError

Occurs when Python encounters invalid indentation.

### 2. TabError

Occurs when tabs and spaces are mixed inconsistently.

### 3. Incorrect Block Boundaries

A statement may be placed inside the wrong block while remaining syntactically valid.

This can lead to incorrect calculations, repeated execution, or skipped operations.

### 4. Unintended Loop Behavior

Incorrect indentation can cause aggregations, reporting operations, or conditional checks to execute at the wrong stage.

### 5. Uninitialized Variables

Variables assigned inside conditional branches may not exist if the relevant conditions are never satisfied.

### Prevention Strategies

- Use consistent four-space indentation.
- Avoid mixing tabs and spaces.
- Keep nested control structures readable.
- Initialize important variables before conditional execution.
- Use automated formatting and static analysis tools.
- Test edge cases and alternative execution paths.


```python
# Portfolio position values

position_values = [12000, 8500, 15500, 9000]

# Incorrect logic:
# The reported value is overwritten at every iteration.

total_value = 0
reported_value = 0

for value in position_values:
    total_value += value
    reported_value = total_value

# This is the final accumulated value, but assigning
# it repeatedly inside the loop is unnecessary.

print(f"Reported Value: ${reported_value:,.2f}")


# Preferred implementation

total_value = 0

for value in position_values:
    total_value += value

# Report only after the complete aggregation
print(f"Final Portfolio Value: ${total_value:,.2f}")
```

## 1.2.6 Practical Application: Multi-Asset Risk Monitoring and Stress Testing

### Objective

Develop a structured portfolio risk-monitoring system that evaluates concentration risk and portfolio sensitivity to hypothetical market shocks.

The implementation focuses on advanced indentation, nested control flow, conditional execution, and hierarchical reporting.

### Functional Requirements

1. Calculate individual asset market values.
2. Aggregate market values by sector.
3. Calculate position and sector weights.
4. Identify concentration-limit breaches.
5. Evaluate multiple market stress scenarios.
6. Calculate portfolio losses under each scenario.
7. Identify the largest loss contributor.
8. Generate risk alerts based on predefined thresholds.
9. Identify the worst-performing scenario.
10. Produce a structured risk-monitoring report.

### Technical Requirements

- Use Python's standard library only.
- Implement nested dictionaries and loops.
- Apply multiple levels of conditional logic.
- Preserve the original portfolio data.
- Maintain consistent code indentation.
- Separate aggregation, risk assessment, and reporting.

### Assumptions

- The portfolio is long-only.
- All prices are denominated in USD.
- Cash remains unchanged under stress.
- Sector shocks are applied uniformly to assets within each sector.
- Portfolio weights are measured relative to total equity, including cash.
- Stress shocks are hypothetical and do not represent forecasts.


```python
# ============================================================
# MULTI-ASSET PORTFOLIO RISK MONITORING SYSTEM
# ============================================================

# 1. Portfolio data

portfolio = {
    "AAPL": {
        "sector": "Technology",
        "shares": 50,
        "price": 190.00
    },
    "MSFT": {
        "sector": "Technology",
        "shares": 20,
        "price": 410.00
    },
    "JPM": {
        "sector": "Financials",
        "shares": 40,
        "price": 205.00
    },
    "TLT": {
        "sector": "Bonds",
        "shares": 80,
        "price": 92.00
    }
}

cash = 5000.00

MAX_POSITION_WEIGHT = 0.30
MAX_SECTOR_WEIGHT = 0.45
MAX_STRESS_LOSS = -0.10


# 2. Position and sector aggregation

position_values = {}
sector_values = {}

for ticker, asset in portfolio.items():

    market_value = asset["shares"] * asset["price"]
    sector = asset["sector"]

    position_values[ticker] = market_value

    if sector not in sector_values:
        sector_values[sector] = 0

    sector_values[sector] += market_value


# 3. Portfolio weights

portfolio_value = (
    sum(position_values.values())
    + cash
)

position_weights = {}
sector_weights = {}

for ticker, market_value in position_values.items():
    position_weights[ticker] = (
        market_value / portfolio_value
    )

for sector, market_value in sector_values.items():
    sector_weights[sector] = (
        market_value / portfolio_value
    )


# 4. Concentration risk assessment

risk_alerts = []

for ticker, weight in position_weights.items():

    if weight > MAX_POSITION_WEIGHT:
        risk_alerts.append(
            f"POSITION BREACH: {ticker} ({weight:.2%})"
        )

for sector, weight in sector_weights.items():

    if weight > MAX_SECTOR_WEIGHT:
        risk_alerts.append(
            f"SECTOR BREACH: {sector} ({weight:.2%})"
        )


# 5. Stress scenarios

stress_scenarios = {
    "Technology Sell-Off": {
        "Technology": -0.20,
        "Financials": -0.08,
        "Bonds": 0.03
    },
    "Interest Rate Shock": {
        "Technology": -0.07,
        "Financials": 0.01,
        "Bonds": -0.12
    },
    "Broad Risk-Off": {
        "Technology": -0.25,
        "Financials": -0.15,
        "Bonds": -0.06
    }
}

scenario_results = {}


# 6. Nested scenario analysis

for scenario_name, sector_shocks in stress_scenarios.items():

    scenario_pnl = 0
    largest_loss = 0
    largest_loss_asset = None

    for ticker, asset in portfolio.items():

        sector = asset["sector"]

        if sector not in sector_shocks:
            raise ValueError(
                f"Missing shock for sector: {sector}"
            )

        shock = sector_shocks[sector]
        market_value = position_values[ticker]

        asset_pnl = market_value * shock

        scenario_pnl += asset_pnl

        if asset_pnl < largest_loss:
            largest_loss = asset_pnl
            largest_loss_asset = ticker

    scenario_return = scenario_pnl / portfolio_value

    if scenario_return <= MAX_STRESS_LOSS:
        risk_status = "BREACH"

    else:
        risk_status = "ACCEPTABLE"

    scenario_results[scenario_name] = {
        "pnl": scenario_pnl,
        "return": scenario_return,
        "largest_loss_asset": largest_loss_asset,
        "status": risk_status
    }


# 7. Identify worst-performing scenario

worst_scenario = min(
    scenario_results,
    key=lambda name: scenario_results[name]["return"]
)


# 8. Structured risk report

print("=" * 65)
print("MULTI-ASSET PORTFOLIO RISK REPORT")
print("=" * 65)

print(f"Portfolio Value: ${portfolio_value:,.2f}")
print(f"Cash Balance: ${cash:,.2f}")

print("\nPOSITION WEIGHTS")
print("-" * 65)

for ticker, weight in position_weights.items():
    print(f"{ticker:<8} {weight:>8.2%}")

print("\nSECTOR EXPOSURES")
print("-" * 65)

for sector, weight in sector_weights.items():
    print(f"{sector:<15} {weight:>8.2%}")

print("\nCONCENTRATION ALERTS")
print("-" * 65)

if risk_alerts:

    for alert in risk_alerts:
        print(alert)

else:
    print("All concentration limits satisfied.")

print("\nSTRESS TEST RESULTS")
print("-" * 65)

for name, result in scenario_results.items():

    print(f"\nScenario: {name}")

    print(f"Stress P&L: ${result['pnl']:,.2f}")
    print(f"Stress Return: {result['return']:.2%}")

    print(
        f"Largest Loss Contributor: "
        f"{result['largest_loss_asset']}"
    )

    print(f"Risk Status: {result['status']}")

print("\nWORST-CASE SCENARIO")
print("-" * 65)

print(f"Scenario: {worst_scenario}")

print(
    f"Loss: "
    f"${scenario_results[worst_scenario]['pnl']:,.2f}"
)
```

### Analytical Interpretation

**1. Portfolio Concentration**

Technology represents approximately 46.26% of total portfolio equity, exceeding the predefined sector concentration limit of 45%.

This indicates a concentration-limit violation under the selected risk-management policy.

**2. Stress Test Results**

The broad risk-off scenario generates the largest simulated portfolio loss, approximately 15.93% of total portfolio equity.

The technology sell-off scenario also breaches the predefined 10% stress-loss threshold.

**3. Nested Control Flow**

The program uses nested loops to evaluate each asset under each market scenario.

Conditional statements determine the appropriate risk classification and identify major loss contributors.

**4. Code Structure**

Indentation separates portfolio aggregation, concentration assessment, scenario evaluation, and final reporting.

Incorrect placement of a calculation could cause it to execute at the asset level rather than at the scenario level.

**5. Model Limitations**

The results depend entirely on predefined market shocks. The model does not estimate shock probabilities, dynamic correlations, transaction costs, or liquidity effects.

The analysis should be interpreted as a deterministic sensitivity assessment rather than a probabilistic risk forecast.
