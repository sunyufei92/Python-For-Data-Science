## 1.4.1 Physical Lines and Logical Lines

### Physical Lines

A physical line represents a sequence of source-code characters terminated by a line-ending character.

Depending on the operating system, line endings may be represented by:

- Line Feed (`LF`)
- Carriage Return followed by Line Feed (`CRLF`)
- Carriage Return (`CR`)

Python normalizes recognized line-ending conventions during lexical processing.

### Logical Lines

A logical line represents a complete unit of Python source code constructed according to Python's lexical rules.

A logical line may consist of one physical line or multiple physical lines joined together.

### Relationship Between Physical and Logical Lines

A statement can occupy multiple physical lines while remaining a single logical statement.

Python supports two principal line-joining mechanisms:

**1. Implicit Line Joining**

Occurs within parentheses `()`, square brackets `[]`, and curly braces `{}`.

**2. Explicit Line Joining**

Uses a backslash (`\`) at the end of a physical line.

### Lexical Analysis

Before Python executes a program, its source code is processed into tokens.

Relevant token categories include:

- `NAME`: Identifiers and keywords.
- `NUMBER`: Numeric literals.
- `STRING`: String literals.
- `OP`: Operators and delimiters.
- `NEWLINE`: End of a logical line.
- `NL`: A non-terminating newline, including those inside implicitly continued statements.
- `INDENT`: Increase in indentation level.
- `DEDENT`: Decrease in indentation level.

### Advanced Interpretation

Line continuation changes the physical presentation of source code without necessarily changing its logical structure.

However, modifying parentheses or operator grouping can change how expressions are evaluated.

Therefore, formatting and expression semantics must be considered separately.


```python
import io
import tokenize

source_code = """
portfolio_return = (
    0.40 * 0.12
    + 0.35 * 0.08
    + 0.25 * (-0.03)
)
"""

# Tokenize the source without executing it.
tokens = tokenize.generate_tokens(
    io.StringIO(source_code).readline
)

print("LEXICAL TOKEN ANALYSIS")
print("-" * 50)

for token in tokens:

    if token.type in {
        tokenize.NL,
        tokenize.NEWLINE
    }:
        token_name = tokenize.tok_name[token.type]

        print(
            f"Line: {token.start[0]:>2} | "
            f"Token: {token_name:<10} | "
            f"Text: {token.string!r}"
        )
```

### Analytical Interpretation

**1. Physical Line Structure**

The source expression spans several physical lines.

**2. Logical Statement Structure**

The assignment forms a single logical statement because the arithmetic expression is enclosed in parentheses.

**3. Tokenization**

The tokenizer distinguishes between newlines that terminate logical statements and newlines that occur within continued expressions.

**4. Practical Significance**

Understanding these distinctions is useful when developing:

- Python source-code analyzers.
- Automated formatting tools.
- Static analysis utilities.
- Code-generation systems.
- Advanced debugging tools.

The number of physical lines in an expression does not determine the number of executable statements.


## 1.4.2 Implicit Line Continuation

### Definition

Implicit line continuation allows a Python expression to span multiple physical lines when the expression is enclosed in parentheses, brackets, or braces.

Unlike explicit continuation, implicit continuation does not require a backslash.

### Supported Delimiters

**1. Parentheses `()`**

Commonly used for:

- Mathematical expressions.
- Function calls.
- Generator expressions.
- Multi-line conditions.
- Grouped Boolean expressions.

**2. Square Brackets `[]`**

Commonly used for:

- Lists.
- List comprehensions.
- Indexing and slicing expressions.
- Nested sequences.

**3. Curly Braces `{}`**

Commonly used for:

- Dictionaries.
- Sets.
- Dictionary comprehensions.
- Nested configuration structures.

### Important Properties

- Physical line breaks inside implicit continuation do not terminate the logical statement.
- Comments can appear between elements of an implicitly continued expression.
- Continuation indentation does not define a new code block.
- Trailing commas are permitted in many comma-separated structures.
- Parentheses can also control expression grouping and evaluation order.

### Recommended Practice

Implicit continuation should generally be preferred over explicit continuation for long expressions.

Complex expressions should be divided at meaningful computational boundaries rather than at arbitrary character positions.


```python
from math import isclose

# Portfolio configuration

portfolio = {
    "AAPL": {
        "weight": 0.35,
        "expected_return": 0.12,
        "transaction_cost": 0.002,
    },
    "MSFT": {
        "weight": 0.30,
        "expected_return": 0.09,
        "transaction_cost": 0.0015,
    },
    "JPM": {
        "weight": 0.20,
        "expected_return": 0.07,
        "transaction_cost": 0.001,
    },
    "TLT": {
        "weight": 0.15,
        "expected_return": 0.04,
        "transaction_cost": 0.0005,
    },
}

# Validate portfolio weights.
total_weight = sum(
    asset["weight"]
    for asset in portfolio.values()
)

assert isclose(
    total_weight,
    1.0,
    rel_tol=0.0,
    abs_tol=1e-12,
)

# Calculate expected gross portfolio return.
expected_gross_return = sum(
    asset["weight"] * asset["expected_return"]
    for asset in portfolio.values()
)

# Estimate transaction costs for an assumed full
# one-way allocation from cash into the portfolio.
estimated_transaction_cost = sum(
    asset["weight"] * asset["transaction_cost"]
    for asset in portfolio.values()
)

# Calculate expected net return.
expected_net_return = (
    expected_gross_return
    - estimated_transaction_cost
)

# Multi-line conditional expression.
if (
    isclose(total_weight, 1.0, abs_tol=1e-12)
    and expected_net_return > 0.05
    and estimated_transaction_cost < 0.002
):
    portfolio_status = "ACCEPTABLE"
else:
    portfolio_status = "REVIEW REQUIRED"

print(f"Gross Return: {expected_gross_return:.2%}")
print(f"Estimated Cost: {estimated_transaction_cost:.4%}")
print(f"Net Return: {expected_net_return:.2%}")
print(f"Portfolio Status: {portfolio_status}")
```

## 1.4.3 Explicit Line Continuation

### Definition

Explicit line continuation uses a backslash (`\`) at the end of a physical line to join it with the following physical line.

This technique allows a logical statement to span multiple lines without enclosing the expression in parentheses or other grouping delimiters.

### Syntax Rules

1. The backslash must be the final character before the line ending.
2. A backslash used for explicit continuation cannot be followed by an inline comment.
3. Trailing whitespace after the backslash may invalidate the continuation.
4. Explicit continuation cannot be used to split an identifier into multiple tokens.
5. Explicit continuation should generally be avoided when implicit continuation is available.

### Why Explicit Continuation Is Discouraged

Explicit continuation introduces additional syntactic constraints that can make code more fragile.

A minor formatting change, such as inserting whitespace after the backslash, can cause a syntax error.

### Recommended Alternative

Use parentheses to group long expressions.

Parenthesized expressions are generally easier to maintain and support inline comments between components.

### Advanced Consideration

Explicit continuation is a lexical mechanism.

It is fundamentally different from using a backslash escape sequence within a string literal.


```python
# Explicit line continuation

gross_pnl = 12500.0
transaction_costs = 850.0
financing_costs = 420.0

net_pnl_explicit = gross_pnl - transaction_costs - financing_costs

# Preferred: implicit line continuation

net_pnl_implicit = (
    gross_pnl
    - transaction_costs
    - financing_costs
)

print(f"Explicit Result: ${net_pnl_explicit:,.2f}")
print(f"Implicit Result: ${net_pnl_implicit:,.2f}")

assert isclose(
    net_pnl_explicit,
    net_pnl_implicit,
    rel_tol=0.0,
    abs_tol=1e-12,
)
```

```python
# Store invalid examples as strings so that
# the notebook itself remains executable.

invalid_examples = {
    "Missing grouping": (
        "result = 100 +\n"
        "20"
    ),
    "Comment after continuation": (
        "result = 100 + \\\n"
        "# invalid placement\n"
        "20"
    ),
    "Trailing whitespace after backslash": (
        "result = 100 + \\ \n"
        "20"
    ),
}

print("SYNTAX VALIDATION")
print("-" * 50)

for label, source in invalid_examples.items():

    try:
        compile(source, "<example>", "exec")

    except SyntaxError as error:
        print(f"{label}: {type(error).__name__}")

    else:
        print(f"{label}: VALID")
```

## 1.4.4 Multi-line Formatting and PEP 8

### Professional Formatting Conventions

PEP 8 provides recommendations for formatting Python code in a consistent and readable manner.

For multi-line expressions, the primary concern is preserving the logical structure of an expression while minimizing visual ambiguity.

### 1. Vertical Alignment

Continuation lines can align with the opening delimiter when this improves readability.

### 2. Hanging Indentation

A hanging indent places the continuation elements on subsequent lines with a consistent indentation level.

This approach is common in function calls, function definitions, and nested data structures.

### 3. Operator Placement

Python allows binary operators to appear before or after a line break when the syntax supports implicit continuation.

PEP 8 discusses the readability advantages of placing a binary operator at the beginning of the continued line.

### 4. Trailing Commas

Trailing commas are useful in multi-line collections and argument lists because they simplify future modifications and reduce unnecessary version-control differences.

### 5. Function Definitions

Functions with numerous arguments should use structured multi-line signatures.

### 6. Nested Expressions

Deeply nested expressions should be formatted consistently.

When a mathematical expression becomes difficult to interpret, consider extracting intermediate calculations into named variables.

### 7. Line Length

PEP 8 recommends a maximum code line length of 79 characters.

Some projects adopt different limits, such as 88 characters, when using automated formatters.

The selected limit should be applied consistently within the project.

### Key Principle

Formatting should expose the logical structure of an expression rather than obscure it.


```python
from math import isfinite

def calculate_risk_adjusted_return(
    gross_return: float,
    transaction_cost: float,
    financing_cost: float,
    annual_volatility: float,
    *,
    risk_free_rate: float = 0.0,
    minimum_volatility: float = 1e-10,
) -> float:
    """
    Calculate a simplified risk-adjusted return.

    All return and volatility inputs must use
    consistent annualized conventions.
    """
    if (
        not all(
            isfinite(value)
            for value in (
                gross_return,
                transaction_cost,
                financing_cost,
                annual_volatility,
                risk_free_rate,
            )
        )
        or annual_volatility <= minimum_volatility
    ):
        raise ValueError(
            "Inputs must be finite and volatility "
            "must exceed the specified minimum."
        )

    net_return = (
        gross_return
        - transaction_cost
        - financing_cost
    )

    excess_return = (
        net_return
        - risk_free_rate
    )

    return excess_return / annual_volatility


risk_adjusted_return = calculate_risk_adjusted_return(
    gross_return=0.145,
    transaction_cost=0.008,
    financing_cost=0.004,
    annual_volatility=0.185,
    risk_free_rate=0.035,
)

print(
    f"Risk-Adjusted Return: "
    f"{risk_adjusted_return:.4f}"
)
```

## 1.4.5 Expression Grouping and Operator Precedence

### Expression Grouping

Parentheses are used for two related but distinct purposes:

1. Allowing expressions to span multiple physical lines.
2. Explicitly controlling the grouping and evaluation of subexpressions.

Adding or removing parentheses can change a program's mathematical meaning.

### Operator Precedence

Python evaluates expressions according to defined precedence and associativity rules.

Relevant operators include:

- Exponentiation: `**`
- Multiplication and division: `*`, `/`, `//`, `%`
- Addition and subtraction: `+`, `-`
- Comparisons: `<`, `<=`, `>`, `>=`, `==`, `!=`
- Boolean negation: `not`
- Boolean conjunction: `and`
- Boolean disjunction: `or`
- Conditional expressions: `x if condition else y`

### Boolean Evaluation

Python evaluates Boolean expressions using short-circuit semantics.

For `and`, evaluation stops when the result is determined to be false.

For `or`, evaluation stops when the result is determined to be true.

### Financial Computation Risks

Incorrect grouping can lead to:

- Incorrect portfolio returns.
- Miscalculated leverage ratios.
- Invalid risk-adjusted performance metrics.
- Incorrect compounding calculations.
- Incorrect stress-test results.
- Unexpected Boolean classifications.

### Best Practices

- Use explicit parentheses for complex mathematical relationships.
- Separate intermediate terms when their financial interpretation is important.
- Avoid depending on readers to remember complex precedence rules.
- Test calculations against independently derived reference results.
- Prefer readability over excessively compact mathematical expressions.


```python
# ============================================================
# EXPRESSION GROUPING AND FINANCIAL CORRECTNESS
# ============================================================

initial_capital = 100000.0

gross_profit = 12500.0
transaction_costs = 1750.0
financing_costs = 750.0

# Correct: all costs are deducted before
# normalizing the result by initial capital.

correct_net_return = (
    (
        gross_profit
        - transaction_costs
        - financing_costs
    )
    / initial_capital
)

# Incorrect: division applies only to financing costs.
incorrect_net_return = (
    gross_profit
    - transaction_costs
    - financing_costs / initial_capital
)

print(f"Correct Return: {correct_net_return:.2%}")
print(f"Incorrect Return: {incorrect_net_return:.2%}")


# ============================================================
# ADDITIVE VS. MULTIPLICATIVE PRICE SHOCKS
# ============================================================

sector_shock = -0.18
security_shock = -0.03

# Additive approximation
additive_shock = (
    sector_shock
    + security_shock
)

# Sequential multiplicative shock composition
multiplicative_shock = (
    (1 + sector_shock)
    * (1 + security_shock)
    - 1
)

print(f"\nAdditive Shock: {additive_shock:.2%}")
print(f"Multiplicative Shock: {multiplicative_shock:.2%}")


# ============================================================
# MULTI-LINE BOOLEAN VALIDATION
# ============================================================

portfolio_value = 125000.0
gross_exposure = 1.35
maximum_leverage = 1.50

if (
    portfolio_value > 0
    and gross_exposure <= maximum_leverage
    and correct_net_return > 0
):
    validation_status = "PASSED"
else:
    validation_status = "FAILED"

print(f"\nValidation Status: {validation_status}")
```

## 1.4.6 Multi-line Strings and Formatted Reporting

### Multi-line Statements vs. Multi-line Strings

Multi-line statements and multi-line string literals serve different purposes.

A multi-line statement distributes executable syntax across several physical lines.

A multi-line string literal represents textual data that may contain newline characters.

### Triple-Quoted Strings

Triple-quoted strings can span multiple physical lines.

They preserve line breaks as part of the string value.

### Implicit String Literal Concatenation

Adjacent string literals can be concatenated implicitly when placed within a single expression.

This is especially useful for constructing long messages and formatted financial reports.

### Multi-line F-Strings

Formatted string literals can be combined across physical lines using parenthesized expressions.

Each formatted segment can contain its own expressions and format specifications.

### Formatting Considerations

Professional financial reports should account for:

- Numeric precision.
- Currency symbols.
- Percentage formatting.
- Thousand separators.
- Column alignment.
- Explicit newline characters.
- Consistent labels and units.

### Important Distinction

Parenthesized adjacent string literals do not automatically insert newline characters.

Explicit `\n` characters or a deliberate string-joining strategy are required when line breaks should appear in the final output.


```python
from textwrap import dedent

# Portfolio summary data

portfolio_value = 250000.0
net_return = 0.0875
annual_volatility = 0.162
maximum_drawdown = -0.124

# Parenthesized adjacent f-strings
report = (
    "PORTFOLIO PERFORMANCE REPORT\n"
    "========================================\n"
    f"Portfolio Value : ${portfolio_value:>12,.2f}\n"
    f"Net Return      : {net_return:>12.2%}\n"
    f"Annual Volatility: {annual_volatility:>11.2%}\n"
    f"Maximum Drawdown : {maximum_drawdown:>11.2%}\n"
)

print(report)


# Triple-quoted string with indentation normalization
methodology = dedent(
    """
    METHODOLOGY

    1. Returns are expressed in decimal form internally.
    2. Volatility is assumed to be annualized.
    3. Drawdown is calculated relative to running peaks.
    4. Transaction costs are included in net returns.
    """
).strip()

print(methodology)


# Dynamic multi-line report generation
risk_metrics = {
    "Annual Volatility": 0.162,
    "Maximum Drawdown": -0.124,
    "Tracking Error": 0.053,
}

metric_lines = [
    f"{name:<20}: {value:>8.2%}"
    for name, value in risk_metrics.items()
]

dynamic_report = "\n".join(
    [
        "RISK METRICS",
        "-" * 35,
        *metric_lines,
    ]
)

print("\n" + dynamic_report)
```

## 1.4.7 Semantic Preservation and Source Code Analysis

### Semantic Preservation

A formatting refactor is intended to improve source-code readability without changing the observable behavior of the program.

However, changing parentheses, operators, or evaluation order can unintentionally modify the meaning of an expression.

### Abstract Syntax Trees

Python represents parsed source code using an abstract syntax tree (AST).

An AST describes the structural relationships between program elements.

It generally does not preserve every formatting detail from the original source.

### AST-Based Comparison

The `ast` module can be used to compare structural representations of different source-code versions.

If two code fragments produce equivalent AST structures, they often represent the same Python operations.

However, structural equality alone does not prove equivalent behavior in every execution environment.

### Static Analysis

Static analysis examines program structure without executing the program's full runtime behavior.

Relevant techniques include:

- Lexical token inspection.
- Abstract syntax tree parsing.
- Syntax validation.
- Source-code formatting checks.
- Structural comparisons.
- Detection of potentially misleading expressions.

### Important Limitations

AST comparison does not prove mathematical correctness.

It also does not guarantee that code is free from side effects, floating-point issues, or runtime errors.

For financial models, formatting refactoring should be supported by numerical regression tests.

### Recommended Workflow

1. Identify a complex expression.
2. Record its expected behavior.
3. Rewrite the expression with clearer formatting.
4. Compare source-code structure where appropriate.
5. Execute numerical regression tests.
6. Review edge cases and assumptions.


```python
import ast

# Compact source code
original_source = """
portfolio_value = 100000 * (1 - 0.15) + 2500
"""

# Reformatted source code
refactored_source = """
portfolio_value = (
    100000 * (1 - 0.15)
    + 2500
)
"""

# A mathematically different expression
modified_source = """
portfolio_value = (
    100000 * (1 - (0.15 + 2500))
)
"""


def get_ast_structure(source_code: str) -> str:
    """Return a normalized AST representation."""
    parsed = ast.parse(source_code)

    return ast.dump(
        parsed,
        include_attributes=False,
    )


original_ast = get_ast_structure(original_source)
refactored_ast = get_ast_structure(refactored_source)
modified_ast = get_ast_structure(modified_source)

# Formatting changes preserve structural meaning.
assert original_ast == refactored_ast

# Changing mathematical grouping changes the AST.
assert original_ast != modified_ast

print("AST STRUCTURAL VERIFICATION")
print("-" * 50)

print(
    "Formatting-only refactor:",
    original_ast == refactored_ast,
)

print(
    "Modified expression structure:",
    original_ast == modified_ast,
)


# Numerical regression verification
original_namespace = {}
refactored_namespace = {}

exec(original_source, {}, original_namespace)
exec(refactored_source, {}, refactored_namespace)

assert isclose(
    original_namespace["portfolio_value"],
    refactored_namespace["portfolio_value"],
    rel_tol=0.0,
    abs_tol=1e-12,
)

print("\nNumerical regression test passed.")
```

## 1.4.8 Practical Application: Advanced Portfolio Stress Testing and Risk Attribution

### Objective

Develop a deterministic portfolio stress-testing engine capable of evaluating multiple market scenarios, calculating asset-level P&L contributions, and identifying portfolio risk-limit violations.

The implementation emphasizes professional multi-line expression formatting, numerical correctness, and maintainable analytical code.

### Functional Requirements

1. Represent multiple financial assets using structured data models.
2. Validate portfolio holdings and scenario parameters.
3. Calculate individual position market values.
4. Calculate portfolio and sector exposures.
5. Apply sector-level and security-specific price shocks.
6. Combine shocks using multiplicative compounding.
7. Calculate asset-level scenario P&L contributions.
8. Aggregate scenario P&L and stressed portfolio values.
9. Identify concentration-limit violations.
10. Identify scenarios breaching the portfolio stress-loss threshold.
11. Rank scenarios according to their financial impact.
12. Validate core numerical relationships using automated assertions.

### Technical Requirements

- Python 3.10 or later.
- Python standard library only.
- Structured multi-line expressions.
- Parenthesized mathematical formulas.
- Multi-line function signatures.
- Multi-line function calls.
- Nested data structures.
- Explicit numerical validation.
- Type annotations.
- Consistent PEP 8 formatting.
- Deterministic and reproducible results.

### Financial Assumptions

- All holdings are long-only.
- Assets are denominated in USD.
- Cash remains unchanged under market stress.
- Sector and security shocks are applied sequentially.
- Shocks represent instantaneous percentage price changes.
- Transaction costs and financing costs are excluded.
- All holdings are valued at a single point in time.
- The analysis is deterministic rather than probabilistic.

### Numerical Definitions

For asset i under scenario s:

Combined Shock:

    R(i,s) = (1 + SectorShock(i,s))
             * (1 + SecurityShock(i,s)) - 1

Position P&L:

    PnL(i,s) = MarketValue(i) * R(i,s)

Portfolio Scenario P&L:

    PnL(s) = Sum(PnL(i,s))

Portfolio Scenario Return:

    Return(s) = PnL(s) / InitialPortfolioValue

Stressed Portfolio Value:

    StressedValue(s) = InitialPortfolioValue + PnL(s)


```python
# ============================================================
# ADVANCED PORTFOLIO STRESS TESTING ENGINE
# ============================================================

from dataclasses import dataclass
from math import isclose, isfinite


@dataclass(frozen=True)
class Holding:
    ticker: str
    sector: str
    units: int
    price: float


@dataclass(frozen=True)
class Scenario:
    name: str
    sector_shocks: dict[str, float]
    ticker_shocks: dict[str, float]


# Portfolio configuration
holdings = [
    Holding("AAPL", "Technology", 80, 185.0),
    Holding("MSFT", "Technology", 50, 410.0),
    Holding("JPM", "Financials", 60, 195.0),
    Holding("XOM", "Energy", 90, 105.0),
    Holding("TLT", "Bonds", 120, 92.0),
]

cash_balance = 8500.0

MAX_POSITION_WEIGHT = 0.25
MAX_SECTOR_WEIGHT = 0.40
MAX_STRESS_LOSS = 0.12


# Market scenarios
scenarios = [
    Scenario(
        name="Technology Correction",
        sector_shocks={
            "Technology": -0.18,
            "Financials": -0.05,
            "Energy": -0.06,
            "Bonds": 0.04,
        },
        ticker_shocks={
            "AAPL": -0.03,
        },
    ),
    Scenario(
        name="Inflation Repricing",
        sector_shocks={
            "Technology": -0.10,
            "Financials": -0.07,
            "Energy": 0.08,
            "Bonds": -0.13,
        },
        ticker_shocks={
            "MSFT": -0.02,
        },
    ),
    Scenario(
        name="Systemic Selloff",
        sector_shocks={
            "Technology": -0.26,
            "Financials": -0.20,
            "Energy": -0.15,
            "Bonds": -0.10,
        },
        ticker_shocks={
            "AAPL": -0.05,
            "JPM": -0.04,
        },
    ),
]


def is_finite_number(value: object) -> bool:
    """Check whether a value is a finite real number."""
    return (
        isinstance(value, (int, float))
        and not isinstance(value, bool)
        and isfinite(value)
    )


def validate_inputs(
    holdings: list[Holding],
    scenarios: list[Scenario],
    cash: float,
) -> None:
    """Validate portfolio and scenario configurations."""

    if not holdings or not scenarios:
        raise ValueError(
            "Holdings and scenarios cannot be empty."
        )

    if (
        not is_finite_number(cash)
        or cash < 0
    ):
        raise ValueError(
            "Cash must be finite and nonnegative."
        )

    tickers = set()
    sectors = set()

    for holding in holdings:

        if not holding.ticker or not holding.sector:
            raise ValueError(
                "Ticker and sector are required."
            )

        if holding.ticker in tickers:
            raise ValueError("Duplicate ticker.")

        if (
            not isinstance(holding.units, int)
            or isinstance(holding.units, bool)
            or holding.units <= 0
        ):
            raise ValueError(
                "Units must be positive integers."
            )

        if (
            not is_finite_number(holding.price)
            or holding.price <= 0
        ):
            raise ValueError(
                "Prices must be finite and positive."
            )

        tickers.add(holding.ticker)
        sectors.add(holding.sector)

    scenario_names = set()

    for scenario in scenarios:

        if (
            not scenario.name
            or scenario.name in scenario_names
        ):
            raise ValueError(
                "Scenario names must be unique and nonempty."
            )

        scenario_names.add(scenario.name)

        if not sectors.issubset(scenario.sector_shocks):
            raise ValueError(
                "Missing a required sector shock."
            )

        if not set(scenario.ticker_shocks).issubset(tickers):
            raise ValueError(
                "Unknown ticker in scenario."
            )

        for shock in (
            *scenario.sector_shocks.values(),
            *scenario.ticker_shocks.values(),
        ):
            if (
                not is_finite_number(shock)
                or shock < -1.0
            ):
                raise ValueError(
                    "Shocks must be finite and at least -100%."
                )


print("Portfolio and scenario configurations created.")
```

```python
def evaluate_stress(
    holdings: list[Holding],
    scenarios: list[Scenario],
    cash: float,
    *,
    max_position_weight: float = 0.25,
    max_sector_weight: float = 0.40,
    max_loss: float = 0.12,
) -> dict:
    """
    Evaluate deterministic portfolio stress scenarios.

    Returns portfolio valuation, concentration alerts,
    and scenario-level risk attribution results.
    """
    validate_inputs(
        holdings=holdings,
        scenarios=scenarios,
        cash=cash,
    )

    for limit in (
        max_position_weight,
        max_sector_weight,
        max_loss,
    ):
        if (
            not is_finite_number(limit)
            or not 0 <= limit <= 1
        ):
            raise ValueError(
                "Risk limits must lie between 0 and 1."
            )

    # Calculate individual market values.
    position_values = {
        holding.ticker: holding.units * holding.price
        for holding in holdings
    }

    portfolio_value = (
        sum(position_values.values())
        + cash
    )

    if (
        not isfinite(portfolio_value)
        or portfolio_value <= 0
    ):
        raise ValueError(
            "Portfolio value must be finite and positive."
        )

    # Aggregate sector exposures.
    sector_values = {}

    for holding in holdings:

        position_value = position_values[holding.ticker]

        sector_values[holding.sector] = (
            sector_values.get(holding.sector, 0.0)
            + position_value
        )

    # Identify concentration-limit violations.
    alerts = [
        f"POSITION: {ticker}"
        for ticker, value in position_values.items()
        if value / portfolio_value > max_position_weight
    ]

    alerts += [
        f"SECTOR: {sector}"
        for sector, value in sector_values.items()
        if value / portfolio_value > max_sector_weight
    ]

    scenario_results = {}

    # Evaluate individual scenarios.
    for scenario in scenarios:

        contributions = {
            holding.ticker: (
                position_values[holding.ticker]
                * (
                    (
                        1 + scenario.sector_shocks[holding.sector]
                    )
                    * (
                        1 + scenario.ticker_shocks.get(
                            holding.ticker,
                            0.0,
                        )
                    )
                    - 1
                )
            )
            for holding in holdings
        }

        scenario_pnl = sum(
            contributions.values()
        )

        scenario_return = (
            scenario_pnl
            / portfolio_value
        )

        worst_contributor = min(
            contributions,
            key=contributions.get,
        )

        if contributions[worst_contributor] >= 0:
            worst_contributor = None

        scenario_results[scenario.name] = {
            "pnl": scenario_pnl,
            "return": scenario_return,
            "stressed_value": (
                portfolio_value
                + scenario_pnl
            ),
            "worst_contributor": worst_contributor,
            "contributions": contributions,
            "breach": (
                scenario_return < -max_loss
            ),
        }

    return {
        "portfolio_value": portfolio_value,
        "position_values": position_values,
        "sector_values": sector_values,
        "alerts": alerts,
        "results": scenario_results,
    }


# Execute portfolio analysis.
report = evaluate_stress(
    holdings=holdings,
    scenarios=scenarios,
    cash=cash_balance,
    max_position_weight=MAX_POSITION_WEIGHT,
    max_sector_weight=MAX_SECTOR_WEIGHT,
    max_loss=MAX_STRESS_LOSS,
)

print("Portfolio risk analysis completed.")
```

```python
# ============================================================
# STRUCTURED PORTFOLIO STRESS REPORT
# ============================================================

portfolio_value = report["portfolio_value"]
results = report["results"]

print("=" * 70)
print("ADVANCED PORTFOLIO STRESS TESTING REPORT")
print("=" * 70)

print(
    f"Portfolio Value: "
    f"${portfolio_value:,.2f}"
)

print("\nCONCENTRATION ALERTS")
print("-" * 70)

if report["alerts"]:

    for alert in report["alerts"]:
        print(alert)

else:
    print("No concentration violations.")


print("\nSCENARIO COMPARISON")
print("-" * 70)

ranked_scenarios = sorted(
    results.items(),
    key=lambda item: item[1]["return"],
)

for name, result in ranked_scenarios:

    print(
        f"{name:<25}"
        f"P&L: ${result['pnl']:>12,.2f}  "
        f"Return: {result['return']:>8.2%}"
    )


print("\nDETAILED SCENARIO ATTRIBUTION")
print("-" * 70)

for name, result in ranked_scenarios:

    print(f"\nScenario: {name}")

    for ticker, contribution in sorted(
        result["contributions"].items(),
        key=lambda item: item[1],
    ):
        print(
            f"  {ticker:<8}"
            f"${contribution:>12,.2f}"
        )

    print(
        f"  Largest Loss Contributor: "
        f"{result['worst_contributor']}"
    )

    print(
        f"  Risk Limit Breach: "
        f"{result['breach']}"
    )


worst_name, worst_result = ranked_scenarios[0]

print("\nWORST-CASE SCENARIO")
print("-" * 70)

print(
    f"Scenario: {worst_name}\n"
    f"Stress P&L: ${worst_result['pnl']:,.2f}\n"
    f"Stress Return: {worst_result['return']:.2%}\n"
    f"Stressed Value: "
    f"${worst_result['stressed_value']:,.2f}"
)
```

### Analytical Interpretation

**1. Multi-line Expression Structure**

The implementation demonstrates how complex mathematical operations can be distributed across multiple physical lines while preserving their logical structure.

Parenthesized expressions improve the readability of financial calculations involving nested arithmetic and conditional logic.

**2. Mathematical Grouping**

The stress-testing engine combines sector and security-specific shocks multiplicatively.

Explicit parentheses make the compounding assumptions visible and reduce the risk of incorrectly applying additive approximations.

**3. Computational Decomposition**

The portfolio analysis separates valuation, exposure aggregation, concentration assessment, stress simulation, and reporting.

This structure makes the analytical workflow easier to verify and maintain.

**4. Financial Risk Attribution**

Scenario-level P&L is decomposed into individual position contributions.

The sum of position contributions is verified against the portfolio-level P&L.

**5. Numerical Verification**

Automated assertions verify valuation identities, return calculations, scenario aggregation, and predefined reference results.

Numerical verification is essential because syntactically valid code may still contain financially incorrect calculations.

**6. Static vs. Runtime Correctness**

Lexical analysis and abstract syntax trees can help inspect and compare program structures.

However, static structural checks cannot replace execution-based regression testing.

**7. Performance Considerations**

Multi-line formatting does not inherently improve computational efficiency.

Performance depends on algorithm design, data structures, memory access patterns, and the operations being performed.

**8. Model Limitations**

The stress-testing engine is deterministic and relies on manually specified price shocks.

It does not estimate scenario probabilities, dynamic correlations, market liquidity, or portfolio trading costs.

The implementation uses floating-point arithmetic and is intended for analytical research rather than accounting-grade financial settlement.

**9. Software Engineering Implications**

Professional multi-line formatting improves readability, code review, and maintainability.

It also reduces the likelihood of accidental changes to mathematical expressions during future model extensions.
