### 1.1.1 Python Syntax and Statements

Python programs consist of statements and expressions.

- **Statement:** An instruction that performs an action, such as assigning a variable or executing a function.
- **Expression:** A combination of values, variables, and operators that evaluates to a result.
- **Case Sensitivity:** Python distinguishes between uppercase and lowercase identifiers.
- **Statement Termination:** Most statements end at a newline without requiring a semicolon.
- **Dynamic Typing:** Variable types are determined at runtime rather than explicitly declared.

**Key Distinction:**

An expression produces a value, whereas a statement performs an operation. Some statements contain expressions.


```python
# Variable assignment statements
stock_price = 185.50
number_of_shares = 100

# Arithmetic expression
market_value = stock_price * number_of_shares

# Output statement
print(f"Market Value: ${market_value:,.2f}")

# Python is case-sensitive
price = 100
Price = 200

print(price)
print(Price)

# Multiple assignment
asset_a, asset_b = "AAPL", "MSFT"

print(asset_a, asset_b)
```

### 1.1.2 Indentation and Code Blocks

Python uses indentation to define the structure and scope of code blocks.

A code block is a group of statements executed together under a particular condition, loop, function, or class.

**Key Rules:**

- Indentation is mandatory for defining nested code blocks.
- Four spaces per indentation level is the standard convention.
- Statements within the same block must use consistent indentation.
- A colon (`:`) typically introduces an indented block.
- Incorrect indentation may raise `IndentationError` or `TabError`.

**Nested Blocks:**

Python supports multiple levels of indentation, allowing complex decision-making and program structures.


```python
# Portfolio risk classification

portfolio_return = -0.08
risk_threshold = -0.05
annual_volatility = 0.28

if portfolio_return < risk_threshold:
    if annual_volatility > 0.25:
        risk_level = "High Risk"
    else:
        risk_level = "Moderate Risk"
else:
    risk_level = "Low Risk"

print(f"Portfolio Return: {portfolio_return:.2%}")
print(f"Annual Volatility: {annual_volatility:.2%}")
print(f"Risk Classification: {risk_level}")
```

### 1.1.3 Comments and Docstrings

Comments and docstrings improve code readability, documentation, and maintainability.

**1. Single-Line Comments**

A single-line comment begins with the `#` symbol. Python ignores the remainder of the line.

**2. Multi-line Comments**

Python does not have a dedicated multi-line comment syntax. Multiple consecutive `#` comments are commonly used.

**3. Docstrings**

Docstrings are string literals used to document modules, functions, classes, and methods.

They are typically enclosed in triple quotes (`"""`) and appear as the first statement inside the object being documented.

Unlike comments, docstrings can be accessed programmatically through the `__doc__` attribute or the `help()` function.

**Best Practices:**

- Explain the purpose or reasoning behind non-obvious code.
- Avoid comments that merely repeat what the code does.
- Use docstrings to document the behavior, parameters, and return values of functions.
- Keep documentation consistent with the implementation.


```python
# Single-line comment: define portfolio parameters
initial_capital = 10000
annual_return = 0.08

# Estimate portfolio value after one year.
# This assumes a fixed annual return of 8%.
future_value = initial_capital * (1 + annual_return)

print(f"Future Value: ${future_value:,.2f}")


# Function documentation using a docstring
def calculate_profit(revenue, cost):
    """
    Calculate profit as the difference between
    revenue and cost.

    Parameters
    ----------
    revenue : float
        Total revenue.
    cost : float
        Total cost.

    Returns
    -------
    float
        Profit or loss.
    """
    return revenue - cost


profit = calculate_profit(15000, 12000)

print(f"Profit: ${profit:,.2f}")

# Access the function's documentation
print(calculate_profit.__doc__)
```

### 1.1.4 Multi-line Statements

Python allows statements to span multiple physical lines, improving readability when working with long expressions or complex calculations.

**1. Implicit Line Continuation**

Expressions enclosed in parentheses `()`, square brackets `[]`, or curly braces `{}` can span multiple lines without special characters.

**2. Explicit Line Continuation**

A backslash (`\`) can be used to continue a statement onto the next line.

However, implicit line continuation is generally preferred because it is clearer and less error-prone.

**Best Practices:**

- Use parentheses for long mathematical expressions.
- Break complex calculations into logically meaningful components.
- Maintain consistent indentation for continued expressions.


```python
# Calculate a weighted portfolio return

portfolio_return = (
    0.40 * 0.12
    + 0.35 * 0.08
    + 0.25 * (-0.03)
)

print(f"Portfolio Return: {portfolio_return:.2%}")
```

### 1.1.5 Python Coding Conventions (PEP 8)

PEP 8 is the official style guide for Python code. It provides recommendations for writing readable, consistent, and maintainable programs.

Following established coding conventions is particularly important in collaborative software development and production-level data science projects.

**Key Guidelines:**

**1. Naming Conventions**

- Variables and functions: `snake_case`
- Classes: `PascalCase`
- Constants: `UPPER_CASE`
- Modules and packages: short, lowercase names

**2. Indentation**

Use four spaces per indentation level.

**3. Line Length**

PEP 8 recommends limiting code lines to 79 characters, although some projects adopt longer limits.

**4. Whitespace**

Use spaces around binary operators and after commas.

**5. Imports**

Place imports at the beginning of a Python module and organize them into standard library, third-party, and local imports.

**6. Code Readability**

Prefer descriptive identifiers, consistent formatting, and simple expressions over unnecessarily complex implementations.


```python
# PEP 8 naming conventions

MAX_POSITION_WEIGHT = 0.25

portfolio_value = 100000
asset_weight = 0.20

# Calculate the allocated capital
allocated_capital = portfolio_value * asset_weight

# Check position limits
if asset_weight <= MAX_POSITION_WEIGHT:
    print("Position weight is within the limit.")
else:
    print("Position weight exceeds the limit.")

print(f"Allocated Capital: ${allocated_capital:,.2f}")
```

### 1.1.6 Practical Application: Advanced Portfolio Risk Analytics

**Objective**

Develop a Python-based portfolio analytics engine that evaluates investment performance, portfolio concentration, and exposure to hypothetical market shocks.

**Key Tasks**

1. Calculate individual position values and unrealized profits or losses.
2. Calculate total portfolio value and overall return.
3. Determine position weights and sector exposures.
4. Identify concentration risk and risk-limit violations.
5. Simulate portfolio performance under multiple market stress scenarios.
6. Identify the worst-performing scenario.

**Technical Concepts**

- Nested data structures
- Conditional statements and nested loops
- Functions and reusable calculations
- Input validation and exception handling
- Dictionary comprehensions
- Sorting and aggregation
- Multi-line expressions
- Formatted financial reporting

**Assumptions**

- All positions are long-only.
- Prices and portfolio values are denominated in USD.
- No deposits, withdrawals, dividends, or transaction costs occur.
- Each sector receives a uniform percentage price shock within a given scenario.
- Stress scenarios are hypothetical and are not probabilistic forecasts.


```python
# ============================================================
# ADVANCED PORTFOLIO RISK & STRESS TESTING ENGINE
# ============================================================

# 1. Portfolio configuration

portfolio = {
    "AAPL": {
        "sector": "Technology",
        "shares": 50,
        "entry_price": 170.00,
        "market_price": 192.00
    },
    "MSFT": {
        "sector": "Technology",
        "shares": 30,
        "entry_price": 360.00,
        "market_price": 410.00
    },
    "JPM": {
        "sector": "Financials",
        "shares": 60,
        "entry_price": 185.00,
        "market_price": 205.00
    },
    "XOM": {
        "sector": "Energy",
        "shares": 80,
        "entry_price": 105.00,
        "market_price": 112.00
    },
    "TLT": {
        "sector": "Bonds",
        "shares": 100,
        "entry_price": 95.00,
        "market_price": 92.00
    }
}

cash_balance = 7500.00

# Portfolio risk constraints
MAX_POSITION_WEIGHT = 0.22
MAX_SECTOR_WEIGHT = 0.35


# 2. Validate inputs and calculate position values

position_values = {}
position_pnl = {}
sector_values = {}

for ticker, asset in portfolio.items():

    shares = asset["shares"]
    entry_price = asset["entry_price"]
    market_price = asset["market_price"]
    sector = asset["sector"]

    if shares <= 0 or entry_price <= 0 or market_price <= 0:
        raise ValueError(f"Invalid position data: {ticker}")

    cost_basis = shares * entry_price
    market_value = shares * market_price

    position_values[ticker] = market_value
    position_pnl[ticker] = market_value - cost_basis

    sector_values[sector] = (
        sector_values.get(sector, 0)
        + market_value
    )

if cash_balance < 0:
    raise ValueError("Cash balance cannot be negative.")


# 3. Portfolio valuation and performance

invested_value = sum(position_values.values())

portfolio_value = invested_value + cash_balance

total_cost_basis = sum(
    asset["shares"] * asset["entry_price"]
    for asset in portfolio.values()
)

initial_capital = total_cost_basis + cash_balance

total_pnl = portfolio_value - initial_capital
total_return = total_pnl / initial_capital


# 4. Position weights and concentration

position_weights = {
    ticker: value / portfolio_value
    for ticker, value in position_values.items()
}

sector_weights = {
    sector: value / portfolio_value
    for sector, value in sector_values.items()
}

# Herfindahl concentration index for invested assets
portfolio_hhi = sum(
    (value / invested_value) ** 2
    for value in position_values.values()
)

largest_position = max(
    position_weights,
    key=position_weights.get
)


# 5. Portfolio risk-limit checks

risk_violations = []

for ticker, weight in position_weights.items():
    if weight > MAX_POSITION_WEIGHT:
        risk_violations.append(
            f"{ticker}: position limit exceeded"
        )

for sector, weight in sector_weights.items():
    if weight > MAX_SECTOR_WEIGHT:
        risk_violations.append(
            f"{sector}: sector limit exceeded"
        )


# 6. Define hypothetical stress scenarios

stress_scenarios = {
    "Equity Sell-Off": {
        "Technology": -0.12,
        "Financials": -0.09,
        "Energy": -0.07,
        "Bonds": 0.05
    },
    "Inflation Shock": {
        "Technology": -0.08,
        "Financials": -0.04,
        "Energy": 0.06,
        "Bonds": -0.10
    },
    "Broad Risk-Off": {
        "Technology": -0.22,
        "Financials": -0.18,
        "Energy": -0.16,
        "Bonds": -0.12
    }
}

scenario_results = {}

for scenario_name, sector_shocks in stress_scenarios.items():

    if not sector_values.keys() <= sector_shocks.keys():
        raise ValueError(
            f"Missing sector shocks: {scenario_name}"
        )

    scenario_pnl = sum(
        sector_value * sector_shocks[sector]
        for sector, sector_value in sector_values.items()
    )

    scenario_results[scenario_name] = {
        "pnl": scenario_pnl,
        "return": scenario_pnl / portfolio_value,
        "stressed_value": portfolio_value + scenario_pnl
    }


# 7. Identify worst-case scenario

worst_scenario = min(
    scenario_results,
    key=lambda name: scenario_results[name]["return"]
)


# 8. Generate portfolio risk report

print("=" * 60)
print("ADVANCED PORTFOLIO RISK REPORT")
print("=" * 60)

print(f"Portfolio Value: ${portfolio_value:,.2f}")
print(f"Unrealized P&L: ${total_pnl:,.2f}")
print(f"Portfolio Return: {total_return:.2%}")
print(f"Concentration HHI: {portfolio_hhi:.4f}")
print(f"Largest Position: {largest_position}")

print("\nPOSITION EXPOSURES")
print("-" * 60)

for ticker, weight in sorted(
    position_weights.items(),
    key=lambda item: item[1],
    reverse=True
):
    print(
        f"{ticker:<8}"
        f"Weight: {weight:>7.2%}  "
        f"P&L: ${position_pnl[ticker]:>10,.2f}"
    )

print("\nSECTOR EXPOSURES")
print("-" * 60)

for sector, weight in sorted(
    sector_weights.items(),
    key=lambda item: item[1],
    reverse=True
):
    print(f"{sector:<15} {weight:>8.2%}")

print("\nRISK LIMIT CHECKS")
print("-" * 60)

if risk_violations:
    for violation in risk_violations:
        print(f"WARNING: {violation}")
else:
    print("All concentration limits satisfied.")

print("\nSTRESS TEST RESULTS")
print("-" * 60)

for name, result in scenario_results.items():
    print(
        f"{name:<20}"
        f"P&L: ${result['pnl']:>10,.2f}  "
        f"Return: {result['return']:>8.2%}"
    )

print("\nWORST-CASE SCENARIO")
print("-" * 60)

print(f"Scenario: {worst_scenario}")
print(
    f"Stressed Value: "
    f"${scenario_results[worst_scenario]['stressed_value']:,.2f}"
)
```

#### Analytical Interpretation

**Portfolio Performance**

The portfolio has a positive unrealized return of approximately 7.28%, indicating an increase in market value relative to its aggregate acquisition cost and unchanged cash balance.

**Concentration Risk**

Technology represents approximately 36.59% of total portfolio equity, exceeding the predefined sector concentration limit of 35%.

This indicates that the portfolio has relatively high exposure to technology-sector price movements under the selected classification.

**Stress Testing**

The broad risk-off scenario produces the largest simulated portfolio loss, approximately 15.99% of current portfolio equity.

Stress testing helps quantify sensitivity to specified market shocks. However, these results depend on the assumed shocks and do not represent statistical confidence intervals or Value at Risk estimates.

**Limitations**

- The model excludes transaction costs, financing costs, and taxes.
- It assumes prices change immediately according to the specified sector shocks.
- It does not model correlations dynamically, liquidity constraints, or nonlinear instruments.
- The concentration index measures asset concentration, not total market risk.
- The analysis uses a single valuation snapshot rather than a historical time series.
